## 00 – Setup & Data Check

This notebook verifies that the environment and raw Amazon dataset are ready for downstream analysis.

It will:
- Load the raw Amazon reviews/products data.
- Run basic sanity checks on shape, columns, dtypes, missing values, and duplicates.
- Normalize column names and clean key numeric fields (prices, discounts, ratings, rating counts).
- Expand multi-valued categories and review information into a tidy, row-level structure.
- Save the processed dataset for use in later notebooks (for example, `01_eda.ipynb`).

```python
import numpy as np
import pandas as pd

RAW_PATH = "../data/raw/amazon.csv"
PROCESSED_PATH = "../data/processed/amazon.csv"

df = pd.read_csv(RAW_PATH)
```

```python
def summarize_df(df: pd.DataFrame, name: str = "df") -> None:
    print(f"\n=== SUMMARY: {name} ===")
    print(f"Shape: {df.shape[0]} rows  d7 {df.shape[1]} columns")
    print("\nColumns:", df.columns.tolist())
    print("\nDtypes:")
    print(df.dtypes)
    print("\nMissing values (top 10):")
    print(df.isnull().sum().sort_values(ascending=False).head(10))
    print("\nDuplicate rows:", df.duplicated().sum())


summarize_df(df, name="raw data")
print("\n=== DESCRIBE (head) ===")
print(df.describe(include="all").T.head(10))

print("\n=== INFO ===")
print(df.info())
```

### 3. Column name normalization

```python
print("Before column normalization:")
print(df.columns.tolist())

df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_", regex=False)

print("\nAfter column normalization:")
print(df.columns.tolist())
```

### 4. Numeric and currency cleaning

```python
def clean_numeric_column(df: pd.DataFrame, col: str, pattern: str = r"[^\d.-]") -> None:
    """Utility to strip non-numeric characters and convert a column to numeric."""
    print(f"\n=== Cleaning {col} ===")
    before = df[[col]].head(3)
    print("Before:")
    print(before)

    df[col] = df[col].astype(str).str.replace(pattern, "", regex=True)
    df[col] = pd.to_numeric(df[col], errors="coerce")

    print("After:")
    print(df[[col]].head(3))


clean_numeric_column(df, "discounted_price")
clean_numeric_column(df, "actual_price")
clean_numeric_column(df, "discount_percentage")
```

## Separating categories

```python

# dirty_spend_mask = df['discount_percentage'].astype(str).str.contains(r'\%')
# df["category_list"] = df["category"].astype(str).str.split("|")
# df_exploded = df.explode("category_list")

df["category"] = df["category"].str.split("|")
df = df.explode("category")
# df_exploded["category"].value_counts().head(10)
# df_exploded.head(10)


cols = ["user_id", "user_name", "review_id", "review_title", "review_content"]

# 1) Turn "{A,B,C}" into ["A","B","C"] in each of the columns
for c in cols:
    df[c] = (df[c]
             .astype("string")
             .str.strip()
             .str.split(",")         # split into list
            )

# 2) Trim whitespace inside items
for c in cols:
    df[c] = df[c].apply(lambda xs: [x.strip() for x in xs] if isinstance(xs, list) else xs)

# 3) (optional but important) check aligned lengths
lens = df[cols].map(len)
bad = df[lens.nunique(axis=1) != 1]
print("Rows with mismatched list lengths:", len(bad))

# 4) Zip aligned lists into list-of-tuples per row
df["__zipped__"] = df[cols].apply(lambda r: list(zip(*r)), axis=1)

# 5) Explode to multiple rows
df = df.explode("__zipped__").dropna(subset=["__zipped__"]).copy()

# 6) Expand tuple into columns
df[cols] = pd.DataFrame(df["__zipped__"].tolist(), index=df.index)

# 7) Cleanup
df = df.drop(columns="__zipped__")

df
```
