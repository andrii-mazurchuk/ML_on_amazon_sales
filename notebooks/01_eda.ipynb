```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/amazon.csv")
```

```python
df.head()
```

```python
total_products = df["product_id"].nunique()
total_reviews  = df["review_id"].nunique()
total_users = df["user_id"].nunique()
total_categories = df["category"].nunique()

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Products",
        "Total Reviews",
        "Unique Users",
        "Total Categories"
    ],
    "Value": [
        total_products,
        total_reviews,
        total_users,
        total_categories
    ]
})

kpi_summary
```

```python
t_df = df.drop_duplicates(["product_id"])

mean_actual_price = t_df["actual_price"].mean()
median_actual_price = t_df["actual_price"].median()

mean_discounted_price = t_df["discounted_price"].mean()
median_discounted_price = t_df["discounted_price"].median()

mean_discount = t_df["discount_percentage"].mean()
median_discount = t_df["discount_percentage"].median()

min_price = t_df["actual_price"].min()
max_price = t_df["actual_price"].max()

mean_rating = t_df["rating"].mean()
median_rating = t_df["rating"].median()
std_rating = t_df["rating"].std()

product_level = t_df.drop_duplicates(subset="product_id")

mean_rating_count = product_level["rating_count"].mean()
median_rating_count = product_level["rating_count"].median()

reviews_per_product = t_df.groupby("product_id")["review_id"].count()

mean_reviews_per_product = reviews_per_product.mean()
median_reviews_per_product = reviews_per_product.median()


reviews_per_user = t_df.groupby("user_id")["review_id"].count()

mean_reviews_per_user = reviews_per_user.mean()
median_reviews_per_user = reviews_per_user.median()


executive_kpis = pd.DataFrame({
    "Metric": [
        "Total Products",
        "Total Reviews",
        "Unique Users",
        "Avg Actual Price",
        "Median Actual Price",
        "Avg Discount %",
        "Avg Rating",
        "Avg Reviews per Product",
        "Avg Reviews per User"
    ],
    "Value": [
        total_products,
        total_reviews,
        total_users,
        round(mean_actual_price, 2),
        round(median_actual_price, 2),
        round(mean_discount, 2),
        round(mean_rating, 2),
        round(mean_reviews_per_product, 2),
        round(mean_reviews_per_user, 2)
    ]
})

executive_kpis
```

```python

t_df = t_df.dropna(subset=["rating", "rating_count"])
C = t_df["rating"].mean()
m = t_df["rating_count"].quantile(0.75)
m = max(m, 1)
# t_df["weighted_rating"] = t_df.apply(lambda x: (
#     (x["rating_count"] / (x["rating_count"] + m)) * x["rating"]
#     +
#     (m / (x["rating_count"] + m)) * C
# ), axis=0)

t_df["weighted_rating"] = (
    (t_df["rating_count"] / (t_df["rating_count"] + m)) * t_df["rating"]
    +
    (m / (t_df["rating_count"] + m)) * C
)

top_products = t_df.sort_values(
    "weighted_rating",
    ascending=False
).head(10)
top_products
```
